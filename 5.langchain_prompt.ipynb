## 1.3 结构化输出

```python
system_prompt="""
# 身份
- 您是一个科幻作家，根据用户的要求创建一个太空之都。

# 指令
- 请以JSON数据格式输出，不加任何markdown样式

# 示例
user: 月球的首都是什么？
assistant: {
    "name": "月亮城",
    "location": "月球的正面",
    "vibe": "拥有先进的科技和美丽的景观",
    "economy": "以旅游业和科技研发为主"
}

user: 火星的首都是什么？
assistant: {
    "name": "火星城",
    "location": "火星的赤道附近",
    "vibe": "拥有先进的科技和独特的地理环境",
    "economy": "以矿业和科技研发为主"
}
"""
agent = create_agent(
    model="deepseek-chat",
    tools=[],  # 空工具列表
    system_prompt=system_prompt
)

# 流式调用（修正输入键为 "messages"）
for token, metadata in agent.stream(
    {"messages": [HumanMessage(content="金星的首都是什么？")]},
    stream_mode="messages",
):
    print(token.content, end="", flush=True)
```

## 1.4 基于model的结构化输出
- 定义一个数据类型（基于pydatic）
- 创建智能体，设置输出格式

```python
from pydantic import BaseModel

# 定义一个类，用来封装模型需要输出的数据
class CapitialInfo(BaseModel):
    name: str
    location: str
    vibe: str
    economy: str

system_prompt="""
您是一个科幻作家，根据用户的要求创建一个太空之都
"""

# 我们可以创建智能体时设置结构化输出的格式，langchain会自动将模型的输出解析成我们定义的格式
agent = create_agent(
    model="deepseek-chat",
    tools=[],  # 空工具列表
    system_prompt=system_prompt,
    response_format=CapitialInfo
)

response = agent.invoke({"messages": [HumanMessage(content="金星的首都是什么？")]})

print(response)
```

```python
city = response['structured_response']
city
```

```python
print(f"{city.name}位于{city.location}，拥有{city.vibe}，经济以{city.economy}为主。")
```
